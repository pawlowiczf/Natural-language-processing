```python
from datasets import load_dataset

from dataset_loading import get_tokenizer

tok = get_tokenizer()  # trains on first run (saved to data/tokenizer.json), later loads from disk
print("vocab size:", tok.get_vocab_size())
```

```python
wiki = load_dataset("Salesforce/wikitext", "wikitext-103-raw-v1", split="test")
lines = [t.strip() for t in wiki["text"] if len(t) > 100 and not t.startswith(" =")]
sentences = [l.split(" . ")[0] + " ." for l in lines[:5]]
```

```python
for s in sentences:
    enc = tok.encode(s)
    print(s)
    print(enc.tokens)
    print(enc.ids)
    print(f"{len(s)} chars -> {len(enc.ids)} tokens, round trip ok: {tok.decode(enc.ids) == s}\n")
```
